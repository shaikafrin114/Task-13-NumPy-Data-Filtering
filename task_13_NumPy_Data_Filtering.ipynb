```python

import numpy as np

# Create an array of student marks
marks = np.array([35, 45, 50, 60, 75, 80, 90, 95])

print("Student Marks:")
print(marks)
```

```python

# Filter marks greater than 50
filtered_marks = marks[marks > 50]

print("Marks greater than 50:")
print(filtered_marks)
```

```python

# Values below 50
below_50 = marks[marks < 50]

# Values equal to 50
equal_50 = marks[marks == 50]

# Values above 50
above_50 = marks[marks > 50]

print("Below 50:", below_50)
print("Equal to 50:", equal_50)
print("Above 50:", above_50)
```

```python

condition = marks > 50

print("Boolean condition:")
print(condition)

print("Filtered values:")
print(marks[condition])
```

```python

# Marks from 50 up to, but not including, 90
selected_marks = marks[(marks >= 50) & (marks < 90)]

print("Marks between 50 and 89:")
print(selected_marks)
```

```python

sales = np.array([1200, 2500, 800, 4500, 3200, 6000, 1500])

print("All sales:", sales)

# Sales above 3000
high_sales = sales[sales > 3000]

# Sales below 2000
low_sales = sales[sales < 2000]

print("Sales above 3000:", high_sales)
print("Sales below 2000:", low_sales)
```

```python

import numpy as np

# 1. Create student marks array
marks = np.array([35, 45, 50, 60, 75, 80, 90, 95])

print("NUMPY DATA FILTERING")
print("-" * 30)
print("Original marks:", marks)

# 2. Boolean indexing
condition = marks > 50
print("\nBoolean condition:", condition)

# 3. Filter using different thresholds
print("\nMarks above 50:", marks[marks > 50])
print("Marks below 50:", marks[marks < 50])
print("Marks equal to 50:", marks[marks == 50])
print("Marks >= 75:", marks[marks >= 75])
print("Marks <= 60:", marks[marks <= 60])

# 4. Filter using multiple conditions
print("\nMarks from 50 to below 90:",
      marks[(marks >= 50) & (marks < 90)])

# 5. Sales dataset
sales = np.array([1200, 2500, 800, 4500, 3200, 6000, 1500])

print("\nOriginal sales:", sales)
print("Sales above 3000:", sales[sales > 3000])
print("Sales below 2000:", sales[sales < 2000])

print("\nTask completed successfully!")
```
